```python
def hill_climber(objective, dimensions, num_searches=10, iter=100, step_size=0.1, bounds=(-10, 10)):
    best_result = float('inf')
    all_histories = []

    for search in range(num_searches):
        best_solution = np.random.uniform(bounds[0], bounds[1], dimensions)
        best_value = objective(best_solution)
        history = [best_value]

        for _ in range(iter):
            neighbor = best_solution + np.random.uniform(-step_size, step_size, dimensions)
            neighbor = np.clip(neighbor, bounds[0], bounds[1])

            neighbor_value = objective(neighbor)

            if neighbor_value < best_value:
                best_solution, best_value = neighbor, neighbor_value

            history.append(best_value)

        if history[-1] < best_result:
            best_result = history[-1]
        all_histories.append(history)

    return all_histories, best_result


def greedy_hill_climber(objective, dimensions, iter=100, step_size=0.1, bounds=(-10, 10)):
    best_solution = np.random.uniform(bounds[0], bounds[1], dimensions)
    best_value = objective(best_solution)
    history = [best_value]

    for _ in range(iter):
        best_neighbor_value = best_value
        best_neighbor = best_solution

        for _ in range(10):
            neighbor = best_solution + np.random.uniform(-step_size, step_size, dimensions)
            neighbor = np.clip(neighbor, bounds[0], bounds[1])
            neighbor_value = objective(neighbor)

            if neighbor_value < best_neighbor_value:
                best_neighbor_value = neighbor_value
                best_neighbor = neighbor

        best_solution, best_value = best_neighbor, best_neighbor_value
        history.append(best_value)

    return history, best_value


def objective1(x):
    a = 1
    b = 100
    return (a - x[0])**2 + b * (x[1] - x[0]**2)**2

def objective2(x):
    sum_part = sum(x_i**2 for x_i in x) / 4000
    product_part = np.prod(np.cos(x / np.sqrt(np.arange(1, len(x) + 1))))
    return 1 + sum_part - product_part

def plot_all_histories(all_histories, title):
    plt.figure(figsize=(10, 6))
    for i, history in enumerate(all_histories):
        plt.plot(history, label=f"Search {i + 1}")
    plt.xlabel("Iteration")
    plt.ylabel("Objective Value")
    plt.title(title)
    plt.legend()
    plt.grid()
    plt.show()

histories1, best_result1 = hill_climber(objective1, dimensions=2)
plot_all_histories(histories1, "Objective1 Optimization Histories")
print(f"Best result with multiple: {best_result1}")
histories1, best_result1 = hill_climber(objective1, dimensions=2, num_searches=1, iter=100)
plot_all_histories(histories1, "Objective1 Optimization Histories")
print(f"Best result with single: {best_result1}")

histories2, best_result2 = hill_climber(objective2, dimensions=5)
plot_all_histories(histories2, "Objective2 Optimization Histories")
print(f"Best result with multiple: {best_result2}")
histories2, best_result2 = hill_climber(objective2, dimensions=5, num_searches=1, iter=100)
plot_all_histories(histories2, "Objective2 Optimization Histories")
print(f"Best result with single: {best_result2}")

avgs = [0, 0, 0, 0, 0, 0]
times = [0, 0, 0, 0, 0, 0]

for _ in range(100):
    start_time = time.time()
    _, best_result1 = hill_climber(objective1, dimensions=2, num_searches=1)
    times[0] += time.time() - start_time
    
    start_time = time.time()
    _, best_result2 = hill_climber(objective1, dimensions=2)
    times[1] += time.time() - start_time

    start_time = time.time()
    _, best_result3 = greedy_hill_climber(objective1, dimensions=2)
    times[2] += time.time() - start_time
    
    start_time = time.time()
    _, best_result4 = hill_climber(objective2, dimensions=5, num_searches=1)
    times[3] += time.time() - start_time

    start_time = time.time()
    _, best_result5 = hill_climber(objective2, dimensions=5)
    times[4] += time.time() - start_time

    start_time = time.time()
    _, best_result6 = greedy_hill_climber(objective2, dimensions=5)
    times[5] += time.time() - start_time

    avgs[0] += best_result1
    avgs[1] += best_result2
    avgs[2] += best_result3
    avgs[3] += best_result4
    avgs[4] += best_result5
    avgs[5] += best_result6

avg_results = [avg / 1000 for avg in avgs]
avg_times = [t / 1000 for t in times]

print(f"\n\nAverages of best results:")
print(f"hill_climber 2 dims: {avg_results[0]}")
print(f"hill_climber 2 dims 3 workers: {avg_results[1]}")
print(f"greedy_hill_climber 2 dims: {avg_results[2]}")
print(f"hill_climber 5 dims: {avg_results[3]}")
print(f"hill_climber 5 dims 3 workers: {avg_results[4]}")
print(f"greedy_hill_climber 5 dims: {avg_results[5]}")

print(f"\nAverage times:")
print(f"hill_climber 2 dims: {avg_times[0]}")
print(f"hill_climber 2 dims 3 workers: {avg_times[1]}")
print(f"greedy_hill_climber 2 dims: {avg_times[2]}")
print(f"hill_climber 5 dims: {avg_times[3]}")
print(f"hill_climber 5 dims 3 workers: {avg_times[4]}")
print(f"greedy_hill_climber 5 dims: {avg_times[5]}")
    
```

<h2> Task 3. Implement Simulated Annealing </h2> with following options for "cooling" the temperature:

*   Linear Cooling Schedule: Tk = T0 − α⋅k, where T0 is initial temperature, α is a constant and k is iteration number.
*   Logarithmic Cooling Schedule: Tk = (T0)/(log(k+1)), where T0 is initial temperature and k is iteration number
*   Exponential Cooling Schedule: Tk = T0 ⋅ B^k, where T0 is initial temperature, k is iteration number and 0 < β < 1 is cooling rate
*   Adaptive Cooling Schedule: An adaptive temperature schedule adjusts the cooling rate based on the algorithm's performance, such as the acceptance rate of new solutions. The idea is to slow down the cooling process if the algorithm is still finding better solutions and to speed it up if the algorithm is stagnating
*   Custom cooling schedule: A custom temperature schedule can be designed to fit specific problem characteristics or based on empirical observations. Get creative in this one. It does not have to work better, but have an explanation of your idea.


```python
def rastrigin(x):
    A = 10
    n = len(x)
    return A * n + sum(x_i**2 - A * np.cos(2 * np.pi * x_i) for x_i in x)

def linear_cooling(temp, cooling_rate, iteration, acceptance_rate, custom_data):
    return temp - cooling_rate * iteration

def logarithmic_cooling(temp, cooling_rate, iteration, acceptance_rate, custom_data):
    return temp /  (math.log(iteration + 1, 20 - (19 * cooling_rate)) + 1e-9)

def exponential_cooling(temp, cooling_rate, iteration, acceptance_rate, custom_data):
    return temp - math.pow(1 + cooling_rate, iteration)

def adaptive_cooling(temp, cooling_rate, iteration, acceptance_rate, custom_data):
    if acceptance_rate == 0:
        return temp /  (math.log(iteration - 1, 12 - (10 * cooling_rate)) + 1e-9)
    else:
        try:
            return temp /  (math.log(iteration + 1, 12 - (10 * cooling_rate)) + 1e-9)
        except:
            return 99999999

#https://www.scirp.org/journal/paperinformation?paperid=78834
#jednostavnija verzija toga, ako je vise puta u inneru nasao dobro rjesenje vise smanjuje temp, ako ne onda ga manje smanjuje
#sa intervalom [0.6-0.99] dosta cesto zapinjao pa je promjenjen u [0.8, 0.99]
#i taj je prebrzo convergeao kada bi bio na rubu i imao previse prihvatljivih odabira pa je promjenjen u neki hibrid adaptivnog sa intervalom [0.6, 1.1] jer sa 0.9 pre dugo traje a ovako convergea brze od logaritma
def custom_cooling(temp, cooling_rate, iteration, acceptance_rate, custom_data):
    a = 0.8
    b = 1.1
    return custom_data[0] * (a + (b - a) * (1 - custom_data[1]))

def simulated_annealing(objective_function, initial_temp, cooling_function, cooling_rate, num_iteration, inner_iterations, dimensions):
    current_solution = np.random.uniform(-5, 5, dimensions)
    best_solution = current_solution
    best_fitness = objective_function(best_solution)
    path, temperatures = [], []

    temp = initial_temp
    for iteration in range(num_iteration):
        custom_data = [temp, 0]
        acceptance_rate = 0
        current_fitness = objective_function(current_solution)
        for inner in range(inner_iterations):
            neighbor_solution = current_solution + np.random.uniform(-0.2, 0.2, size=current_solution.shape)
            neighbor_solution = np.clip(neighbor_solution, -5, 5)
            neighbor_fitness = objective_function(neighbor_solution)

            if neighbor_fitness < current_fitness:
                prob = 1.0
            else:
                prob = np.exp((current_fitness - neighbor_fitness) / (temp + 1e-99))

            if np.random.rand() < prob:
                if prob != 1:
                    custom_data[1] += 1
                acceptance_rate = 1
                current_solution = neighbor_solution
                current_fitness = neighbor_fitness
                if neighbor_fitness < best_fitness:
                    best_solution = neighbor_solution
                    best_fitness = neighbor_fitness

        path.append(current_solution.copy())
        temperatures.append(temp)
        custom_data[1] /= inner_iterations

        temp = cooling_function(initial_temp, cooling_rate, iteration, acceptance_rate, custom_data)


    return best_solution, best_fitness, path

def visualize_simulated_annealing(objective_function, cooling_functions, initial_temp, cooling_rate, num_iterations, inner_iterations, dimensions):
    x_vals = np.linspace(-5.12, 5.12, 100)
    y_vals = np.linspace(-5.12, 5.12, 100)
    X, Y = np.meshgrid(x_vals, y_vals)
    Z = np.array([[objective_function([x, y]) for x in x_vals] for y in y_vals])

    plt.figure(figsize=(30, 30))

    for i, (cooling_name, cooling_function) in enumerate(cooling_functions.items()):
        if cooling_name == "Linear":
            cooling_rate = 0.3
        else:
            cooling_rate = 0.9
        best_solution, best_fittness, path = simulated_annealing(
            objective_function, initial_temp, cooling_function, cooling_rate, num_iterations, inner_iterations, dimensions
        )

        print(f"{cooling_name}\nBest solution: {best_solution}\nBest fitness: {best_fittness}")

        path = np.array(path)

        plt.subplot(3, 2, i + 1)
        plt.contourf(X, Y, Z, levels=50, cmap='viridis')
        plt.colorbar(label='Objective Function Value')
        plt.plot(path[:, 0], path[:, 1], color='red', marker='o', markersize=2, label='Path')
        plt.title(f'{cooling_name} Cooling')
        plt.xlabel('X1')
        plt.ylabel('X2')
        plt.legend()

    plt.tight_layout()
    plt.show()

cooling_functions = {
    "Linear": linear_cooling,
    "Logarithmic": logarithmic_cooling,
    "Exponential": exponential_cooling,
    "Adaptive": adaptive_cooling,
    "Custom": custom_cooling
}

visualize_simulated_annealing(
    rastrigin, cooling_functions, initial_temp=100, cooling_rate=0.9, num_iterations=300, inner_iterations=10, dimensions=2
)
```

<h2> Task 5. Implement Particle Swarm Optimization algorithm </h2>
Select a multidimensional function to optimize (e.g., Rosenbrock function, Rastrigin function, or Griewank function).


1.  Define a class for the particle that includes properties such as position, velocity, personal best position, and personal best value.
2.  Initialize a swarm of particles with random positions and velocities within defined boundaries.
3.  Implement PSO formula for position update
4.  Create a running loop, which perform the optimization


```python
def rastrigin(x):
    return 10 * len(x) + sum((xi ** 2 - 10 * np.cos(2 * np.pi * xi)) for xi in x)

def rosenbrock(x):
    return sum(100 * (x[i+1] - x[i]**2)**2 + (1 - x[i])**2 for i in range(len(x) - 1))

def griewank(x):
    sum_term = sum(xi**2 for xi in x) / 4000
    prod_term = np.prod([np.cos(xi / np.sqrt(i+1)) for i, xi in enumerate(x)])
    return 1 + sum_term - prod_term


class Particle:
    def __init__(self, bounds, dimensions):
        self.position = np.random.uniform(bounds[0], bounds[1], dimensions)
        self.velocity = np.random.uniform(-1, 1, dimensions)
        self.best_position = self.position.copy()
        self.best_score = float('inf')
        self.positions = [self.position.copy()]

    def update_velocity(self, global_best_position, w, c1, c2):
        if w == -1:
            w = 1.1 - (global_best_position / self.best_position)
        r1, r2 = np.random.rand(2)
        cognitive = c1 * r1 * (self.best_position - self.position)
        social = c2 * r2 * (global_best_position - self.position)
        self.velocity = w * self.velocity + cognitive + social

    def move(self, bounds):
        self.position += self.velocity
        self.position = np.clip(self.position, bounds[0], bounds[1])
        self.positions.append(self.position.copy())


class Swarm:
    def __init__(self, func, num_particles, bounds, dimensions, w=-1, c1=0.5, c2=0.5, goal=0.001):
        self.func = func
        self.num_particles = num_particles
        self.particles = [Particle(bounds, dimensions) for _ in range(num_particles)]
        self.global_best_position = np.random.uniform(bounds[0], bounds[1], dimensions)
        self.global_best_score = float('inf')
        self.w = w
        self.c1 = c1
        self.c2 = c2
        self.bounds = bounds
        self.dimensions = dimensions
        self.goal = goal

    def optimize(self, num_iterations):
        for iteration in range(num_iterations):
            if self.global_best_score <= self.goal:
                return self.global_best_position, self.global_best_score, iteration
            for particle in self.particles:
                score = self.func(particle.position)

                if score < particle.best_score:
                    particle.best_score = score
                    particle.best_position = particle.position.copy()

                if score < self.global_best_score:
                    self.global_best_score = score
                    self.global_best_position = particle.position.copy()

            for particle in self.particles:
                particle.update_velocity(self.global_best_position, self.w, self.c1, self.c2)
                particle.move(self.bounds)

        return self.global_best_position, self.global_best_score, iteration


def plot_results(func, best_position, best_score, swarm, title):
    if len(best_position) == 2:
        x = np.linspace(swarm.bounds[0], swarm.bounds[1], 100)
        y = np.linspace(swarm.bounds[0], swarm.bounds[1], 100)
        X, Y = np.meshgrid(x, y)
        Z = np.array([func([xi, yi]) for xi, yi in zip(X.ravel(), Y.ravel())]).reshape(X.shape)

        plt.contourf(X, Y, Z, levels=50, cmap='viridis')
        plt.colorbar()

        for particle in swarm.particles:
            trail = np.array(particle.positions)
            plt.plot(trail[:, 0], trail[:, 1], 'o-', markersize=2, linewidth=1, alpha=0.7)

        plt.plot(best_position[0], best_position[1], 'r*', markersize=10, label=f"Optimal Solution: {best_score:.2f}")
        plt.title(title)
        plt.legend()
        plt.show()
    else:
        print(f"Optimization result: {best_position} with score: {best_score:.2f}")


if __name__ == "__main__":
    num_particles = 10
    num_iterations = 500
    dimensions = 2
    bounds = (-10, 10)

    functions = {"Rastrigin": rastrigin, "Rosenbrock": rosenbrock, "Griewank": griewank}
    for name, func in functions.items():
        swarm = Swarm(func, num_particles, bounds, dimensions, w=0.5, c1=0.5, c2=0.5, goal=0.1)
        best_position, best_score, iterations_done = swarm.optimize(num_iterations)
        plot_results(func, best_position, best_score, swarm, f"{name} function - Best Score: {best_score:.2f} Iterations: {iterations_done}")
```

```python
class AntColony:
    def __init__(self, distances, ants=5, iters=100, decay=0.5, alpha=1, beta=1, start=0):
        self.distances = np.where(distances == 0, np.inf, distances)
        self.pheromones = np.ones(self.distances.shape) / len(distances)
        self.ants = ants
        self.iterations = iters
        self.decay = decay
        self.alpha = alpha
        self.beta = beta
        self.start = start
        
    #complexity => n : prolazi kroz liste velicine n
    def select_city(self, pheromones, distances, visited):
        pheromones = np.copy(pheromones)
        pheromones[list(visited)] = 0
        #valja zapamtiti da je 0**0=1 jer inace se ponavljaju gradovi
        row = np.where(pheromones == 0, 0, (pheromones ** self.alpha)) * ((1.0 / distances) ** self.beta)
        
        if np.sum(row) == 0: 
            row += 1e-10
            
        norm_row = row / row.sum()
        return np.random.choice(range(len(self.distances)), p=norm_row)
    
    #complexity => n^2 : n puta radi select_city koji ima O(n)
    def generate_path(self, start):
        path = []
        visited = set()
        visited.add(start)
        prev = start
        for i in range(len(self.distances) - 1):
            move = self.select_city(self.pheromones[prev], self.distances[prev], visited)
            path.append((prev, move))
            prev = move
            visited.add(prev)
        path.append((prev, start))
        return path
        
    #complexity => n : ima move-ova kolko i gradova
    def calculate_distance(self, path):
        distance = 0
        for move in path:
            distance += self.distances[move]
        return distance
    
    #complexity => a * n^2 : poziva generate_path a puta koliko je i mrava
    def generate_all_paths(self):
        all_paths = []
        for _ in range(self.ants):
            path = self.generate_path(self.start)
            all_paths.append((path, self.calculate_distance(path)))
        return all_paths
    
    #complexity => a * n : onoliko puta koliko je mrava prolazi kroz listu velicine broja gradova
    def spread_pheromones(self, all_paths):
        for path, cost in all_paths:
            for move in path:
                i, j = move 
                self.pheromones[i, j] += 1.0 / cost
        
    #complexity => i * a * n^2 : koristi sve funkcije i radi ih i puta
    def run(self):
        global_shortest = (None, float('inf'))
        patience = 0
        num_iter = 0
        for i in range(self.iterations):
            if patience > 15:
                break
            all_paths = self.generate_all_paths()
            self.pheromones *= (1 - self.decay)
            self.spread_pheromones(all_paths)
            #kada decay=1 onda ostaju vrijednosti samo iz prosle iteracije pa se vraca natrag na posjecene gradove iz nekog razloga, neposjeceni gradi misli da je posjetio jer je pheromone 0
            self.pheromones = np.where(self.pheromones == 0, 1e-20, self.pheromones)
            current_shortest = min(all_paths, key=lambda x: x[1])
            if current_shortest[1] < global_shortest[1]:
                global_shortest = current_shortest
                patience = 0
            else:
                patience += 1
            num_iter = i
                
        return global_shortest, num_iter
        
```

```python
def tournament_selection(population, fitness, k=3):
    selected = random.sample(range(len(population)), k)
    best = max(selected, key=lambda i: fitness[i])
    return population[best]

def roulette_wheel_selection(population, fitness):
    total_fitness = sum(fitness)
    if total_fitness == 0:
        return population[np.random.choice(len(population))]
    probabilities = [f / total_fitness for f in fitness]
    return population[np.random.choice(len(population), p=probabilities)]

def single_point_crossover(parent1, parent2):
    point = random.randint(1, len(parent1) - 1)
    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]
    return child1, child2

def two_point_crossover(parent1, parent2):
    point1, point2 = sorted(random.sample(range(1, len(parent1)), 2))
    child1 = parent1[:point1] + parent2[point1:point2] + parent1[point2:]
    child2 = parent2[:point1] + parent1[point1:point2] + parent2[point2:]
    return child1, child2

def uniform_crossover(parent1, parent2):
    child1, child2 = [], []
    for i in range(len(parent1)):
        if random.random() < 0.5:
            child1.append(parent1[i])
            child2.append(parent2[i])
        else:
            child1.append(parent2[i])
            child2.append(parent1[i])
    return child1, child2

def gaussian_mutation(chromosomes, mutation_rate):
    for i in range(len(chromosomes)):
        if random.random() < mutation_rate:
            chromosomes[i] += np.random.normal()
    return chromosomes

def flip_bit_mutation(chromosomes, mutation_rate):
    for i in range(len(chromosomes)):
        if random.random() < mutation_rate:
            chromosomes[i] = 1 - chromosomes[i]
    return chromosomes

def genetic_algorithm(fitness_function, num_generations, population_size, chromosome_length, mutation_rate, crossover_method, mutation_method, selection_method, elitism_count):
    population = [np.random.randint(2, size=chromosome_length).tolist() for _ in range(population_size)]
    all_time_fitness = 0
    all_time_individual = 0
    
    for generation in range(num_generations):
        fitness = [fitness_function(individual) for individual in population]

        best_idx = np.argmax(fitness)
        best_individual = population[best_idx]
        best_fitness = fitness[best_idx]
        
        if best_fitness > all_time_fitness:
            all_time_fitness = best_fitness
            all_time_individual = best_individual

        elite_indices = np.argsort(fitness)[-elitism_count:][::-1]
        elites = [population[i] for i in elite_indices]

        new_population = elites.copy()  

        while len(new_population) < population_size:
            parent1 = selection_method(population, fitness)
            parent2 = selection_method(population, fitness)
            offspring1, offspring2 = crossover_method(parent1, parent2)
            offspring1 = mutation_method(offspring1, mutation_rate)
            offspring2 = mutation_method(offspring2, mutation_rate)
            new_population.extend([offspring1, offspring2])

        population = new_population[:population_size]

    return all_time_fitness, all_time_individual

rastrigin_dimensions = 10
rastrigin_population_size = 20
rastrigin_generations = 50
mutation_rate = 0.1
elitism_count = 2

best_rastrigin_solution, best_rastrigin_fitness = genetic_algorithm(
    fitness_function=rastrigin_high_dimen,
    num_generations=rastrigin_generations,
    population_size=rastrigin_population_size,
    chromosome_length=rastrigin_dimensions,
    mutation_rate=mutation_rate,
    crossover_method=uniform_crossover,
    mutation_method=gaussian_mutation,
    selection_method=roulette_wheel_selection,
    elitism_count=elitism_count,
)



print("Best Rastrigin Solution:", best_rastrigin_solution)
print("Best Rastrigin Fitness:", best_rastrigin_fitness)

knapsack_population_size = 20
knapsack_generations = 50

best_knapsack_solution, best_knapsack_fitness = genetic_algorithm(
    fitness_function=fitness,
    num_generations=knapsack_generations,
    population_size=knapsack_population_size,
    chromosome_length=len(knapsack_items),
    mutation_rate=mutation_rate,
    crossover_method=single_point_crossover,
    mutation_method=flip_bit_mutation,
    selection_method=tournament_selection,
    elitism_count=elitism_count,
)

print("Best Knapsack Solution:", best_knapsack_solution)
print("Best Knapsack Fitness:", best_knapsack_fitness)
```

```python
# Define the objective function
def quadratic_objective_function(x):
    return (x - 3)**2 + 1

def quadratic_objective_function2(x):
    return x**2

def rastrigin2d(x):
    A = 100
    return A + x**2 - A * np.cos(2 * np.pi * x)

def custom_function(x):
    return -np.cos(10 * x)**2 * np.exp(-x**2)

# Define the numerical gradient estimation
def numerical_gradient(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

def gradient_descent(f, starting_point, learning_rate, num_iterations, h=1):
    x = starting_point
    x_history = []  # To store the history of x values, for visuzalizaton
    f_history = []  # To store the history of function values, for visuzalizaton


    for _ in range(num_iterations):
        x_history.append(x)
        f_history.append(f(x))
        grad = numerical_gradient(f, x, h)
        x = x - learning_rate * grad

    return x, f_history, x_history

final_x, f_values, x_values = gradient_descent(rastrigin2d, 2, 0.03, 300)

print(f"Solution: {final_x}")
print(f"Value: {f_values[-1]}")

x_range = np.linspace(-5, 5, 300)
y_range = [rastrigin2d(x) for x in x_range]

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, label="f(x) = x^2", color="blue")
plt.plot(x_values, f_values, color="red", label="Gradient Descent Path", marker="o", linestyle="-")
plt.title("Function f(x) = x^2 with Gradient Descent Path")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()

final_x, f_values, x_values = gradient_descent(custom_function, 2, 1, 300)

print(f"Solution: {final_x}")
print(f"Value: {f_values[-1]}")

x_range = np.linspace(-5, 5, 300)
y_range = [custom_function(x) for x in x_range]

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, label="f(x) = x^2", color="blue")
plt.plot(x_values, f_values, color="red", label="Gradient Descent Path", marker="o", linestyle="-")
plt.title("Function f(x) = x^2 with Gradient Descent Path")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def pitagora(a, b):
    return math.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2)

def get_customers_in_ride(array, customers):
    array = list(array)
    filtered_customers = [(array[i], customers[i]) for i in range(len(array)) if array[i] != 0]
    sorted_customers = sorted(filtered_customers, key=lambda x: x[0])
    
    return [customer for _, customer in sorted_customers]


def complete_ride(ride, vehicle):
    if len(ride) != 0:
        time = 0
        total_distance = 0
        capacity = vehicle['capacity']
        demand_fullfilled = 0
        curr_location = depot
        for customer in ride:
            total_distance += pitagora(curr_location, customer[2])
            early, late = customer[4]
            
            if capacity >= customer[1]:
                if time < early:
                    time += early + customer[3]
                    capacity -= customer[1]
                    demand_fullfilled += (customer[1] / early)
                elif early <= time <= late:
                    time += customer[3]
                    capacity -= customer[1]
                    demand_fullfilled += customer[1]
                else:
                    time += customer[3]
                    capacity -= customer[1]
                    demand_fullfilled += (customer[1] * 0.4 * (late / time))

        return demand_fullfilled * (demand_fullfilled / (total_distance + pitagora(curr_location, depot)))
    return 0

def fitness_function(individual, vehicles, customers):
    score = 0
    for i in range(len(vehicles)):
        score += complete_ride(get_customers_in_ride(individual[i], customers), vehicles[i])
    return score

def generate_population(customer_size, chromosome_length, population_size, max_customers):
    population = []
    for _ in range(population_size):
        fleet = []
        avaliable_customers = [0] * 20
        for vehicle in range(chromosome_length):
            num_elements = random.randint(1, max_customers)
            numbers_to_fill = random.sample(range(1, 21), 20)
            array = [0] * 20
            j = 1
            route_length = 0
            for i in range(1, num_elements+1):
                if avaliable_customers[numbers_to_fill.index(i)] == 0:
                    if i == 1:
                        start = depot
                    else:
                        start = customers[numbers_to_fill.index(i - 1)][2]
                    if route_length + pitagora(start, customers[numbers_to_fill.index(i)][2]) + pitagora(customers[numbers_to_fill.index(i)][2], depot) < routing_rules['max_distance_per_route']:
                        route_length += pitagora(start, customers[numbers_to_fill.index(i)][2])
                        array[numbers_to_fill.index(i)] = j
                        j += 1
                        avaliable_customers[numbers_to_fill.index(i)] = 1
            fleet.append(array)
        population.append(fleet)
    return population

def roulette_wheel_selection(population, fitness):
    total_fitness = sum(fitness)
    if total_fitness == 0:
        return population[np.random.choice(len(population))]
    probabilities = [f / total_fitness for f in fitness]
    return population[np.random.choice(len(population), p=probabilities)]

def normalize_offspring(offspring):
    result = []
    for chromosome in offspring:
        non_zero_values = {value: index+1 for index, value in enumerate(sorted(set(chromosome) - {0}))}
        normalized_array = [non_zero_values.get(value, 0) for value in chromosome]
        result.append(normalized_array)
    return result

def uniform_crossover(parent1, parent2):
    child1, child2 = [], []
    for i in range(len(parent1)):
        if random.random() < 0.5:
            child1.append(parent1[i])
            child2.append(parent2[i])
        else:
            child1.append(parent2[i])
            child2.append(parent1[i])
    return normalize_offspring(child1), normalize_offspring(child2)

def flip_bit_mutation(chromosomes, mutation_rate):
    for i in range(len(chromosomes)):
        for j in range(len(chromosomes[i])):
            n = max(chromosomes[i])
            if n == 5:
                break
            if random.random() < mutation_rate:
                if chromosomes[i][j] == 0:
                    gene_occupied = False
                    for k in range(len(chromosomes)):
                        if chromosomes[k][j] != 0:
                            gene_occupied = True
                    if not gene_occupied:
                        if n > 0:
                            route_length = pitagora(depot, customers[chromosomes[i].index(1)][2])
                            if n >= 2:
                                for l in range(2, n + 1):
                                    route_length += pitagora(customers[chromosomes[i].index(l - 1)][2], customers[chromosomes[i].index(l)][2])
                                route_length += pitagora(customers[chromosomes[i].index(i)][2], customers[j][2])
                                route_length += pitagora(customers[j][2], depot)
                            elif n == 1:
                                route_length += pitagora(customers[chromosomes[i].index(1)][2], customers[j][2])
                                route_length += pitagora(customers[j][2], depot)
                            if route_length <= routing_rules['max_distance_per_route']:
                                chromosomes[i][j] = n + 1
                        else:
                            chromosomes[i][j] = n + 1
                else:
                    value = chromosomes[i][j]
                    chromosomes[i][j] = 0
                    chromosomes[i] = [x - 1 if x > value else x for x in chromosomes[i]]
            
    return chromosomes

def genetic_algorithm_finale(fitness_function, num_generations, population_size, chromosome_length, mutation_rate, crossover_method, mutation_method, selection_method, elitism_count, vehicles, customers, max_customers=5):
    population = generate_population(len(customers), chromosome_length, population_size, max_customers)
    all_time_fitness = 0
    all_time_individual = 0
    gen = 1
    for generation in range(num_generations):
        fitness = [fitness_function(individual, vehicles, customers) for individual in population]

        best_idx = np.argmax(fitness)
        best_individual = population[best_idx]
        best_fitness = fitness[best_idx]
        
        if best_fitness > all_time_fitness:
            all_time_fitness = best_fitness
            all_time_individual = best_individual

        elite_indices = np.argsort(fitness)[-elitism_count:][::-1]
        elites = [population[i] for i in elite_indices]

        new_population = elites.copy()  
        while len(new_population) < population_size:
            parent1 = selection_method(population, fitness)
            parent2 = selection_method(population, fitness)
            offspring1, offspring2 = crossover_method(parent1, parent2)
            offspring1 = mutation_method(offspring1, mutation_rate)
            offspring2 = mutation_method(offspring2, mutation_rate)
            new_population.extend([offspring1, offspring2])
        population = new_population[:population_size]
        gen += 1
    return all_time_fitness, all_time_individual


mutation_rate = 0.1
elitism_count = 2
population_size = 50
generations = 100

best_fitness, best_solution = genetic_algorithm_finale(
    fitness_function=fitness_function,
    num_generations=generations,
    population_size=population_size,
    chromosome_length=len(vehicles),
    mutation_rate=mutation_rate,
    crossover_method=uniform_crossover,
    mutation_method=flip_bit_mutation,
    selection_method=roulette_wheel_selection,
    elitism_count=elitism_count,
    vehicles=vehicles,
    customers=customers
)

print("Best Solution:", best_solution)
print("Best Fitness:", best_fitness)
```

The results of testing the hyperparameters align with the ones given by simulated annealing
Mutation: in this specific case where there are a lot of options (20 customers) and vehicles can serve only a few it makes sense that the mutation should be real low (adding or removing one or two customers from the route) because making larger changes would cause the vehicles to serve way more customers which they can not, larger mutation rate would be more beneficial if the vehicles had the capacity and time to serve all the customers.
Elitism count: in a case where so few options can be chosen the algorithm greatly benefits from a higher elitism count, using those individuals to remember the viable combinations, but should be cautious with the ration of elitism count to population as it can lead to too few or not enough individuals to explore new combinations
Population and generations: population and generations should be chosen to cover the viable amount of possible combinations, having to many individuals or combinations will lead to repeating the already discovered combinations while having too few will result in not exploring all the good combinations


```python
class AntColonyFinale:
    def __init__(self, customers, depot, vehicles, routing_rules, ants=50, iters=100, decay=0.5, alpha=1, beta=1):
        self.customers = customers
        self.depot = depot
        self.vehicles = vehicles
        self.routing_rules = routing_rules
        self.ants = ants
        self.iterations = iters
        self.decay = decay
        self.alpha = alpha
        self.beta = beta
        self.distances = self.calculate_distances()
        self.pheromones = np.ones(self.distances.shape) / len(customers)
        self.visited = set()

        
    def calculate_distances(self):
        num_customers = len(self.customers)
        distances = np.zeros((num_customers + 1, num_customers + 1))
        for i in range(num_customers):
            for j in range(num_customers):
                if i != j:
                    distances[i + 1][j + 1] = np.sqrt((self.customers[i][2][0] - self.customers[j][2][0])**2 + (self.customers[i][2][1] - self.customers[j][2][1])**2)
        for i in range(num_customers):
            distances[0][i + 1] = np.sqrt((self.depot[0] - self.customers[i][2][0])**2 + (self.depot[1] - self.customers[i][2][1])**2)
            distances[i + 1][0] = distances[0][i + 1]
            
        return distances
    
    def calculate_score(self, prev, capacity, timee):
        score = np.zeros(len(self.distances[prev]))
        score[0] = 0
        for i in range(1, len(self.distances[prev])):
            if self.distances[prev][i] == 0:
                distance = 1e99
            else:
                distance = self.distances[prev][i]
            early, late = self.customers[i - 1][4]
            demand_fullfilled = 0
            
            if capacity >= self.customers[i - 1][1]:
                if timee < early:
                    demand_fullfilled += (self.customers[i - 1][1] / early)
                elif early <= timee <= late:
                    demand_fullfilled += self.customers[i - 1][1]
                else:
                    demand_fullfilled += (self.customers[i - 1][1] * 0.4 * (late / timee)) 
                score[i] = demand_fullfilled * (demand_fullfilled / distance)
            else:
                score[i] = 0
        return score
    
    def adjust_variables(self, capacity, timee, move):
        early, late = self.customers[move - 1][4]
        if timee < early:
            timee += early + self.customers[move - 1][3]
            capacity -= self.customers[move - 1][1]
        elif early <= timee <= late:
            timee += self.customers[move - 1][3]
            capacity -= self.customers[move - 1][1]
        else:
            timee += self.customers[move - 1][3]
            capacity -= self.customers[move - 1][1]
        
        return capacity, timee
    
    def select_customer(self, pheromones, score):
        pheromones = np.copy(pheromones)
        pheromones[list(self.visited)] = 0
        row = np.where(pheromones == 0, 0, (pheromones ** self.alpha)) * (score ** self.beta)
        if np.sum(row) == 0:
            return 0
        norm_row = row / row.sum()
        return np.random.choice(range(len(self.distances)), p=norm_row)
    
    def generate_route(self, vehicle):
        timee = 0
        route = [0]
        capacity = vehicle['capacity']
        prev = 0
        attempts = 0
        while len(route) - 1 < self.routing_rules['max_customers_per_route'] and attempts < 5:
            score = self.calculate_score(prev, capacity, timee)
            move = self.select_customer(self.pheromones[prev], score)
            route_length = 0
            if move == 0:
                break
            if len(route) > 1:
                for i in range(1, len(route)):
                    route_length += pitagora(self.customers[route[i - 1] - 1][2], self.customers[route[i] - 1][2])
                route_length += pitagora(self.customers[route[-1] - 1][2], self.customers[move - 1][2])
                route_length += pitagora(self.customers[move - 1][2], depot)
            if route_length < self.routing_rules['max_distance_per_route']:
                # print("moze")
                # print(route_length)
                # print(move)
                route.append(move)
                attempts = 0
            else:
                # print("predugo")
                # print(move)
                attempts += 1
            capacity, timee = self.adjust_variables(capacity, timee, move)
            prev = move
            self.visited.add(prev)
        route.append(0)
        return route
    
    def calculate_route_score(self, route, vehicle):
        timee = 0
        total_distance = 0
        capacity = vehicle['capacity']
        demand_fullfilled = 0
        curr_location = depot
        for i in route:
            if i != 0:
                total_distance += pitagora(curr_location, self.customers[i - 1][2])
                early, late = self.customers[i - 1][4]
                
                if capacity >= self.customers[i - 1][1]:
                    if timee < early:
                        timee += early + self.customers[i - 1][3]
                        capacity -= self.customers[i - 1][1]
                        demand_fullfilled += (self.customers[i - 1][1] / early)
                    elif early <= timee <= late:
                        timee += self.customers[i - 1][3]
                        capacity -= self.customers[i - 1][1]
                        demand_fullfilled += self.customers[i - 1][1]
                    else:
                        timee += self.customers[i - 1][3]
                        capacity -= self.customers[i - 1][1]
                        demand_fullfilled += (self.customers[i - 1][1] * 0.4 * (late / timee))
                    
        return demand_fullfilled * (demand_fullfilled / (total_distance + pitagora(curr_location, depot)))
    
    def generate_all_routes(self):
        all_routes = []
        for _ in range(self.ants):
            routes = []
            for vehicle in self.vehicles:
                route = self.generate_route(vehicle)
                routes.append((route, self.calculate_route_score(route, vehicle)))
            all_routes.append(routes)
            self.visited = set()
            self.visited.add(0)
        # for routes in all_routes:
        #     print(routes)
        return all_routes
    
    def spread_pheromones(self, all_routes):
        for routes in all_routes:
            for route, cost in routes:
                for i in range(len(route) - 1):
                    self.pheromones[route[i]][route[i + 1]] += 1.0 / cost
    
    def run(self):
        global_best = None
        patience = 0
        num_iter = 0
        for i in range(self.iterations):
            # if patience > 15:
            #     break
            all_routes = self.generate_all_routes()
            self.pheromones *= (1 - self.decay)
            self.spread_pheromones(all_routes)
            self.pheromones = np.where(self.pheromones == 0, 1e-99, self.pheromones)
            current_best = max(all_routes, key=lambda routes: sum(route[1] for route in routes))
            
            if global_best is None or sum(route[1] for route in current_best) > sum(route[1] for route in global_best):
                global_best = current_best
                patience = 0
            else:
                patience += 1
            num_iter = i
        
        return [route[0] for route in global_best], sum(route[1] for route in global_best), num_iter
    

aco = AntColonyFinale(customers, depot, vehicles, routing_rules)
best_solution, best_fitness, num_iter = aco.run()
print("Best solution:", best_solution)
print("Best fitness:", best_fitness)
print("Iterations:", num_iter)
```

```python
#https://www.youtube.com/watch?v=ySN5Wnu88nE&t=737s&ab_channel=Computerphile
#modificirano s obzirom da su sve udaljenosti od koraka do koraka u labirintu 1

#pop i push => O(log n) https://stackoverflow.com/questions/38806202/whats-the-time-complexity-of-functions-in-heapq-library

#heapify inace O(n) al je 1 jer samo dodajem start

#O(1)
def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

#O(n * log n)
def a_star(maze, setlist = 1, start=None, end=None):
    rows, cols = len(maze), len(maze[0])
    if start == None:
        start = (0, 0)
    if end == None:
        end = (rows - 1, cols - 1)
    open_set = [(heuristic(start, end), start)]

    heapq.heapify(open_set)
    came_from = {}
    g_score = {start: 0}
    f_score = {start: heuristic(start, end)}
    if setlist == 1:
        visited = set()
    else:
        visited = list()
    
    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]
    
    #O(n * log n) n => broj mogucih mjesta u labirintu
    while open_set:
        _, current = heapq.heappop(open_set)
        if setlist == 1:
            visited.add(current)
        else:
            visited.append(current)

        
        #O(n) nije bitan za while loop jer samo jednom na kraju
        if current == end:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start)
            return path[::-1], visited
        
        #O(log n)
        for idx, (dr, dc) in enumerate(directions):
            neighbor = (current[0] + dr, current[1] + dc)
            
            if neighbor in visited:
                continue
            
            neighbour_g_score = g_score[current] + 1

            if 0 <= neighbor[0] < rows and 0 <= neighbor[1] < cols and maze[neighbor[0]][neighbor[1]] == 0:
                
                if neighbour_g_score < g_score.get(neighbor, float('inf')):
                    came_from[neighbor] = current
                    g_score[neighbor] = neighbour_g_score
                    f_score[neighbor] = neighbour_g_score + heuristic(neighbor, end)
                    heapq.heappush(open_set, (f_score[neighbor], neighbor))
    
    return [], visited
```

```python
#O(n)
def select_next_move(candidates, end, exponent=2):
    #O(n)
    f_scores = np.array([1 / (f_score * (heuristic(location, end) ** exponent)) for f_score, location in candidates])
    f_scores[0] = f_scores[0] * exponent
    if len(f_scores) == 1:
        return candidates[0]
    
    #O(n)
    norm_f_scores = f_scores / f_scores.sum()
    
    #O(n)
    chosen_index = np.random.choice(len(candidates), p=norm_f_scores)
    return candidates[chosen_index]

#(n * m * log n) n => broj mjesta u labirintu, m => broj kandidata za odabir
def a_star_metaheuristic(maze, setlist=1, start=None, end=None, no_candidates=3, exponent=2):
    rows, cols = len(maze), len(maze[0])
    if start == None:
        start = (0, 0)
    if end == None:
        end = (rows - 1, cols - 1)
    open_set = [(heuristic(start, end), start)]
    heapq.heapify(open_set)
    came_from = {}
    g_score = {start: 0}
    f_score = {start: heuristic(start, end)}
    if setlist == 1:
        visited = set()
    else:
        visited = list()
    
    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]
    
    #(n * m * log n)
    while open_set:
        i = 0
        candidates = []
        
        #O(m * log n)
        while open_set and i < no_candidates:
            current_f_score, current = heapq.heappop(open_set)
            
            if current == end:
                if setlist == 1:
                    visited.add(current)
                else:
                    visited.append(current)
                path = []
                while current in came_from:
                    path.append(current)
                    current = came_from[current]
                path.append(start)
                return path[::-1], visited
            
            candidates.append((current_f_score, current))
            i += 1
        
        #O(m)
        selected_candidate = select_next_move(candidates, end, exponent)
        current = selected_candidate[1]
        
        #O(m) O(1) ako je set
        candidates.remove(selected_candidate)
        
        #O(m * log n)
        for candidate in candidates:
            heapq.heappush(open_set, candidate)

        if setlist == 1:
            visited.add(current)
        else:
            visited.append(current)
        
        #O(log n)
        for idx, (dr, dc) in enumerate(directions):
            neighbor = (current[0] + dr, current[1] + dc)
            
            if neighbor in visited:
                continue
            
            neighbour_g_score = g_score[current] + 1

            if 0 <= neighbor[0] < rows and 0 <= neighbor[1] < cols and maze[neighbor[0]][neighbor[1]] == 0:
                
                if neighbour_g_score < g_score.get(neighbor, float('inf')):
                    came_from[neighbor] = current
                    g_score[neighbor] = neighbour_g_score
                    f_score[neighbor] = neighbour_g_score + heuristic(neighbor, end)
                    heapq.heappush(open_set, (f_score[neighbor], neighbor))
    
    return [], visited
```

```python
#O(n)
def generate_starting_points(maze, start, end, num_points):
    n = len(maze)
    diagonal = []
    points = []
    num_points -= 2
    #O(n)
    for i in range(1, n):
        if maze[i][i] == 0:
            diagonal.append((i, i))
    
    if num_points >= len(diagonal):
        diagonal.append(start)
        diagonal.append(end)
        return diagonal
    
    if num_points == 0:
        return (start, end)
    
    gap = len(diagonal) // num_points
    points.append(start)
    #O(n - 1)
    for i in range(num_points):
        points.append(diagonal[gap * i])
    points.append(end)
    
    return points

def a_star_multi(maze, setlist=1, num_workers=3, alpha=1, beta=1, start=None, end=None):
    n = len(maze)
    if start is None:
        start = (0, 0)
    if end is None:
        end = (n - 1, n - 1)
    
    #O(m)
    starting_points = generate_starting_points(maze, start, end, num_workers)
    #O(m)
    open_sets = [[(heuristic(sp, starting_points[i + 1]), sp)] for i, sp in enumerate(starting_points[:-1])]
    #O(m)
    for open_set in open_sets:
        heapq.heapify(open_set)
    #O(m)
    came_from = [{} for _ in range(num_workers)]
    #O(m)
    g_scores = [{sp: 0} for sp in starting_points[:-1]]
    #O(m)
    f_scores = [{sp: heuristic(sp, starting_points[i + 1])} for i, sp in enumerate(starting_points[:-1])]
    
    if setlist == 1:
        visited = {}
    else:
        visited = []
    #O(m)  
    executed_workers = [set() for _ in range(num_workers - 1)]
    
    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]

    #O(p log n)
    while open_sets:
        #O(p log n)
        for worker_id, open_set in enumerate(open_sets):
            # time.sleep(1)
            #print(f"Worker: {worker_id}")
            if not open_set:
                #print(f"Worker: {worker_id} has no more moves")
                continue
            
            #O(log n)
            current_f, current = heapq.heappop(open_set)
            #print(f"Worker: {worker_id} has moved to {current} with {current_f}")
            
            #O(p)
            if current == end:
                if worker_id == 0:
                    #print(f"Worker: {worker_id} finished the job")
                    #O(p)
                    return reconstruct_path(came_from[worker_id], current), visited, came_from, num_workers
                else:
                    #print(f"Worker: {worker_id} has reached end")
                    if setlist == 1:
                        visited[current] = worker_id
                    else:
                        visited.append((current, worker_id))
                    open_sets[worker_id] = []
                    continue
                
            next_worker_id = worker_id + 1
            #O(m)
            while next_worker_id < len(open_sets) and not open_sets[next_worker_id]:
                next_worker_id += 1
                #print(f"Worker {worker_id} has selected {next_worker_id} as nex worker id")
            if next_worker_id < len(open_sets):
                next_worker_last_point = open_sets[next_worker_id][-1][1]
                #print(f"Worker {worker_id} has aim at {next_worker_last_point}")
            else:
                next_worker_last_point = end
                #print(f"Worker {worker_id} has aim at end")

            if setlist == 1:
                if current in visited:
                    other_worker_id = visited[current]
                    #print(worker_id)
                    #print(other_worker_id)
                    #print(current)
                    #print(visited)
                    #print(other_worker_id is not None)
                    #print(other_worker_id > worker_id)
                    #print(other_worker_id < len(open_sets))
                    #print(other_worker_id not in executed_workers[worker_id])
                else:
                    other_worker_id = None
            else:
                other_worker_id = next((i for i, (point, _) in enumerate(visited) if point == current), None)
            #print(f"Worker {worker_id} has reached visited by {other_worker_id}")
            
            if other_worker_id is None:
                if setlist == 1:
                    visited[current] = worker_id
                else:
                    visited.append((current, worker_id))
            
            #O(p log n)
            if other_worker_id is not None and other_worker_id > worker_id and other_worker_id < len(open_sets) and other_worker_id not in executed_workers[worker_id]:
                executed_workers[worker_id].add(other_worker_id)
                open_sets[other_worker_id] = []
                #print(f"Worker {worker_id} has lower id")
                #O(p)
                if setlist == 1:
                    last_visited = next(k for k, v in reversed(visited.items()) if v == other_worker_id)
                else:
                    #print([point for point, wid in visited if wid == other_worker_id])
                    last_visited = [point for point, wid in visited if wid == other_worker_id][-1]
                #print(f"Other workers {other_worker_id} last visited is {last_visited}")
                #O(p)
                path_segment = reconstruct_path(came_from[other_worker_id], last_visited, current)
                #(p - 1)
                path_segment = path_segment[1:]
                #print(f"Path to recalculate is {path_segment}")
                if setlist == 1:
                    visited[current] = worker_id
                else:
                    visited.append((current, worker_id))
                
                #O(p log n)
                for point in path_segment:
                    if setlist == 1:
                        visited[point] = worker_id
                    else:
                        visited.append((point, worker_id))
                    for direction in directions:
                        neighbor = (point[0] + direction[0], point[1] + direction[1])
                        if 0 <= neighbor[0] < n and 0 <= neighbor[1] < n and maze[neighbor[0]][neighbor[1]] == 0 and neighbor != current:
                            tentative_g_score = g_scores[worker_id].get(point, float('inf')) + 1
                            if tentative_g_score < g_scores[worker_id].get(neighbor, float('inf')):
                                came_from[worker_id][neighbor] = point
                                g_scores[worker_id][neighbor] = tentative_g_score
                                f_scores[worker_id][neighbor] = tentative_g_score + (alpha * heuristic(neighbor, next_worker_last_point)) + (beta * heuristic(neighbor, end))
                                #O(log n)
                                heapq.heappush(open_set, (f_scores[worker_id][neighbor], neighbor))
                #print(came_from[worker_id])
            else:
                if other_worker_id is not None and 0 <= other_worker_id < len(executed_workers) and other_worker_id < worker_id:
                    #print(f"Worker {worker_id} has higher id than {other_worker_id}")
                    # time.sleep(10)
                    open_sets[worker_id] = []
                    executed_workers[other_worker_id].add(worker_id)
                    continue
            
            #O(log n)
            for direction in directions:
                neighbor = (current[0] + direction[0], current[1] + direction[1])
                if 0 <= neighbor[0] < n and 0 <= neighbor[1] < n and maze[neighbor[0]][neighbor[1]] == 0:
                    if current == end:
                        open_sets[worker_id] = []
                        if worker_id == 0:
                            #print(f"Worker: {worker_id} finished the job")
                            return reconstruct_path(came_from[worker_id], current), visited, came_from, num_workers
                    tentative_g_score = g_scores[worker_id].get(current, float('inf')) + 1
                    if tentative_g_score < g_scores[worker_id].get(neighbor, float('inf')):
                        came_from[worker_id][neighbor] = current
                        g_scores[worker_id][neighbor] = tentative_g_score
                        f_scores[worker_id][neighbor] = tentative_g_score + (alpha * heuristic(neighbor, next_worker_last_point)) + (beta * heuristic(neighbor, end))
                        heapq.heappush(open_set, (f_scores[worker_id][neighbor], neighbor))
            
            # if setlist == 1:
            #     moves = [point for point, wid in visited.items() if wid == worker_id]
            # else:
            #     moves = [point for point, wid in visited if wid == worker_id]  
            #print(f"Moves for worker {worker_id}: {moves}")
    
    return None, visited, came_from, num_workers

#O(n)
def reconstruct_path(came_from, current, start=(0, 0)):
    path = [current]
    #O(n)
    while current in came_from:
        # print(current)
        current = came_from[current]
        path.append(current)
        if current == start:
            path.append(current)
            break
    if start == (0, 0):
        path.append(start)
    #O(n)
    path.reverse()
    return path
```

A* metaheuristic

The implementation uf metaheuristic a* algorithm uses weighted random choice to select one of n best next moves similar to the ant colony choosing the next location. The weights are calculated by the formula "1 / (f_score * (heuristic(location, end) ** exponent))" giving the better chances to moves that have a better f_score while doubling down on the heuristic (distance from the end of the maze) because all the distances in the maze are 1 (a good path will have worse f_score then a random chunk near the beginning or the edges that is probably irrelevant) so the algorithm preferably chooses the solutions further along the way. Algorithm also stops on the first encounter with the end of the maze.

Not exploring as much as the original a* with the emphasis of prioritizing areas closer to the end cause the algorithm to explore less therefore finish in less iterations. Due to added functionalities its time complexity is increased, given that information it logical to assume that the algorithm will not outperform its original counterpart in the smaller mazes where it does not have a chance to open up the gap in the number of iterations to make up the time lost caused by its time complexity, while beating it in the bigger mazes. Parameter candidates defines how many next steps to take into the pool for selecting. More candidates means more choices to select a step closer to the end, therefore a higher number would be beneficial, but too high of the number would increase the time spent by too much while taking into account steps which we would not necessarily want to take into account due to their quality. Exponent is a hyperparameter that determines the amount of heuristic we want to take into account, tests showed that the best results were achieved with exponent 3 and 4 which gives a slight bump to the moves closer to end while still giving the moves which would explore some chances. While not being a parameter in the algorithm, gaps denotes walls removed in the maze which creates more routes the algorithm can take to get to the end, the algorithm achieved the best results with gap being 0 because without walls the a* algorithm can navigate the maze more or less freely making a lot of down and right moves and not winding around the maze which is the case in which the a* metaheuristic would have an edge

Initially list were used to keep track of previous moves to maintain order of moves for visualisation and the tests were made using them, after switching to sets for optimisation, a* blew away its metaheuristic counterpart due to it losing it's advantage in iterating because time complexity went from n to 1.

A* multi is an algorithm that tries to optimise solving a maze by having multiple nodes on the diagonal from start to end. the idea behind it is that each node seeks the following node as well as the end, therefore the idea is to hopefully break the maze into few "smaller ones" as when the node reaches the next on it jumps to the last spot that the node explored. This way the algorithm hopefully ignores the areas that are not of interest.

Opposite to the previous algorithm this one excels in mazes with more gaps which allows the nodes to connect to each-other more easily, due to its complexity it works better with fewer nodes. Alpha hyperparameters gives the best result when set to a higher value, because when it's set to a lower one it is basically just a node solving a maze as A* while also having multiple nodes wasting time and resources, beta works best with a medium to bigger value without the effect as big as alpha since alpha in it self in the end kind of leads to end as the last node will seek the end and so consequentally the others will do the same passively
